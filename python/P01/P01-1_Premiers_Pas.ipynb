### Fiche F01-1 : Chemins de fichiers (`pathlib.Path`)

```python
import pathlib
import tempfile
dossier = pathlib.Path(tempfile.mkdtemp(prefix="meca_")) / "essais"    # le dossier d'essai vit dans un dossier temporaire
dossier.mkdir(parents=True, exist_ok=True)
chemin = dossier / "essai_traction.csv"
# Écriture de mesures de force (N) et d'allongement (mm)
chemin.write_text("25000, 1.2\n32000, 2.5", encoding="utf-8", newline="\n")
taille_ko = chemin.stat().st_size / 1000
print(f"Fichier : {chemin.name} ({chemin.suffix})")
print(f"Taille : {taille_ko:.3f} ko".replace(".", ","))
assert chemin.read_text(encoding="utf-8").startswith("25000")
chemin.unlink()
dossier.rmdir()
```
